```python
import threading
import time

def thread1():
    for i in range(5):
        print(f"print{i}")
        time.sleep(1)

def thread2():
    for i in range(5):
        print(f"second{i}")
        time.sleep(1)
t1=threading.Thread(target=thread1)
t2=threading.Thread(target=thread2)

t1.start()
t2.start()

t1.join()
t2.join()

```

```python
#Prime Number
def prime(n):
    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1

    if count == 2:
        print("Prime Number")
    else:
        print("Not a Prime Number")

prime(7)
```

```python
##factorial

def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result
print(factorial(5))
```

```python
# method overriding
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

obj = Dog()
obj.sound()
```

```python
class Parent:
    def display(self):
        print("Parent method")

class Child(Parent):
    def display(self):
        super().display()
        print("Child method")

obj = Child()
obj.display()
```
