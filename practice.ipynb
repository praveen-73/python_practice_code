```python
# python inheritance example
class Employee:
    def work(self):
        print("Employee is working")


class Developer(Employee):
    def code(self):
        print("Developer is writing Python code")


# Create object of child class
dev = Developer()

dev.work()   # Inherited from Employee
dev.code()   # Developer's own method
```

```python
input = ["hello", "i", "praveen"]
vowels = ['a', 'e', 'i', 'o', 'u']
output = []

for i in input:
    count = 0

    for char in i:
        if char in vowels:
            count = count + 1

    output.append([i, count])

for i in range(len(output)):
    for j in range(i + 1, len(output)):
        if output[i][1] > output[j][1]:
            temp = output[i]
            output[i] = output[j]
            output[j] = temp

print(output)
```

```python
input = "()())"
count = 0

for i in input:
    if i == "(":
        count = count + 1
    elif i == ")":
        count = count - 1

    if count < 0:
        break

if count == 0:
    print("Valid")
else:
    print("Invalid")
```

```python
import math

text = input("Enter a string: ")

length = len(text)
columns = math.ceil(math.sqrt(length))
rows = math.ceil(length / columns)

index = 0

for i in range(rows):
    for j in range(columns):
        if index < length:
            print(text[index], end=" ")
            index = index + 1
        else:
            print("-", end=" ")
    print()
```
