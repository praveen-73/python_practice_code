```python
#Linked list:

#It is linear data structures where elements are connected through a series of nodes
# Eascg node consists of a data field and a pointer (or link) to the next node

```

```python
# python inheritance example
class Employee:
    def work(self):
        print("Employee is working")


class Developer(Employee):
    def code(self):
        print("Developer is writing Python code")


# Create object of child class
dev = Developer()

dev.work()   # Inherited from Employee
dev.code()   # Developer's own method
```

```python
word = "tomorrow"
result = ""

for i in range(len(word)):
    if i == 0:
        result += word[i]
    elif i == 2:
        result += word[i]
    elif i == 4 or i == 5:
        result += word[i]
    elif i == 7:
        result += "****" + word[i]
    else:
        result += "*"

print(result)
```
