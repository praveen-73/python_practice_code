```python
# single linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

current = node1

while current:
    print(current.data)
    current = current.next
```

```python
# Doubly Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.prev = node1

node2.next = node3
node3.prev = node2

current = node1

while current:
    print(current.data)
    current = current.next
```

```python
# Circular Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3
node3.next = node1

current = node1

for i in range(3):
    print(current.data)
    current = current.next
```
