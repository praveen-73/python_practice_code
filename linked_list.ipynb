```python
# single linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

current = node1

while current:
    print(current.data)
    current = current.next
```

```python
# Doubly Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.prev = node1

node2.next = node3
node3.prev = node2

current = node1

while current:
    print(current.data)
    current = current.next
```

```python
# Circular Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3
node3.next = node1

current = node1

for i in range(3):
    print(current.data)
    current = current.next
```

```python
# reverse linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Create nodes
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)
head.next.next.next = Node(40)


# Reverse linked list
prev = None
current = head

while current:
    next_node = current.next
    current.next = prev
    prev = current
    current = next_node

head = prev


# Print reversed list
current = head

while current:
    print(current.data, end=" ")
    current = current.next
```
