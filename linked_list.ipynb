```python
# single linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

current = node1

while current:
    print(current.data)
    current = current.next
```

```python
# Doubly Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.prev = node1

node2.next = node3
node3.prev = node2

current = node1

while current:
    print(current.data)
    current = current.next
```

```python
# Circular Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3
node3.next = node1

current = node1

for i in range(3):
    print(current.data)
    current = current.next
```

```python
# reverse linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Create nodes
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)
head.next.next.next = Node(40)


# Reverse linked list
prev = None
current = head

while current:
    next_node = current.next
    current.next = prev
    prev = current
    current = next_node

head = prev


# Print reversed list
current = head

while current:
    print(current.data, end=" ")
    current = current.next
```

```python
#Toeplitz Matrix

matrix = [
    [1, 2, 3],
    [4, 1, 2],
    [7, 4, 1]
]

for i in range(len(matrix) - 1):
    for j in range(len(matrix[0]) - 1):
        if matrix[i][j] != matrix[i + 1][j + 1]:
            print(False)
            break
    else:
        continue
    break
else:
    print(True)
```

```python
items = ["Pen", "Book", "Bag"]
cost = [10, 50, 200]
selling_price = [15, 70, 250]

total_cost = 0
total_profit = 0

for i in range(len(items)):
    total_cost = total_cost + cost[i]
    profit = selling_price[i] - cost[i]
    total_profit = total_profit + profit

    print(items[i], "Profit:", profit)

average_cost = total_cost / len(items)

print("Total cost:", total_cost)
print("Average cost:", average_cost)
print("Total profit:", total_profit)
```

```python
matrix = [
    [1, 0, 0],
    [4, 5, 0],
    [7, 8, 9]
]

n = len(matrix)
result = True

for i in range(n):
    for j in range(n):
        if j > i:
            if matrix[i][j] != 0:
                result = False

if result:
    print("Lower Triangular Matrix")
else:
    print("Not a Lower Triangular Matrix")
```
